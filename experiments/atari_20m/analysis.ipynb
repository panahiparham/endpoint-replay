# atari_20m — DQN on Atari Pong

Plots `dqn_pong`'s episodic-return curves, one line per seed. (`random_pong` is
dropped for now - see the experiment README. Atari is cluster-scale - this
notebook expects results produced elsewhere; see the experiment README.)

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

from experiment import load_curves, load_runs
from experiment.plotting import seed_grids_for, style


def grid_for(component_db, n=500):
    """A shared [0, T] timestep grid from any run's curve length (T = TOTAL_TIMESTEPS)."""
    df = load_runs(component_db, write_csv=False)
    T = len(load_curves(component_db, df["run_id"][0])["reward"])
    return np.linspace(0, T, n)
```

```python
HERE = Path.cwd()
RESULTS = HERE / "results" if (HERE / "results").exists() else Path("experiments/atari_20m/results")

GRID = grid_for(RESULTS / "dqn_pong.db")
df = load_runs(RESULTS / "dqn_pong.db", write_csv=False)
stack = seed_grids_for(RESULTS / "dqn_pong.db", GRID)
seeds = df["seed"].to_list()

colors = ["tab:blue", "tab:red", "tab:green", "tab:orange"]
fig, ax = plt.subplots(figsize=(9, 6))   # 2:3 height:width
for row, seed, color in zip(stack, seeds, colors):
    ax.plot(GRID, row, color=color, label=f"DQN seed {seed}", linewidth=1)

ax.set_title("DQN on Atari Pong")
ax.legend(loc="lower right", frameon=False)
style(ax, ylim=(-21, 21))   # Pong score range
fig.tight_layout()
plt.show()
```

```python
PLOTS_DIR = RESULTS.parent / "plots"
PLOTS_DIR.mkdir(exist_ok=True)
fig.savefig(PLOTS_DIR / "atari_dqn_pong.pdf", bbox_inches="tight")
print(f"saved plot to {PLOTS_DIR / 'atari_dqn_pong.pdf'}")
```
